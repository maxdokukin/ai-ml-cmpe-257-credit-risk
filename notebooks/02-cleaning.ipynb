# Data Cleaning


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/FICO_uncleaned.csv")
df.info()
```

```python
print(df.shape)
print(df.head())
print(df.describe())
```

```python
df = df.replace([-9], np.nan)
df.dropna(inplace=True)

codes = [-8, -7]
df = df.replace(codes, 999)
df.info()
```

```python
# View number of entries with codes [-9, -8, -7] per column
for code in [-9, -8, -7]:
    print(f"Entries with code {code}:")
    print((df == code).sum())
```

```python
df.columns
```

```python
# Label encoder
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['RiskPerformance'] = le.fit_transform(df['RiskPerformance'])
print(df['RiskPerformance'].value_counts())
```

```python
# Write df to a csv file
df.to_csv("../data/processed/FICO_cleaned.csv", index=False)
```

## Alt approach: Let XGBoost handle NaNs and don't remove those rows

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/FICO_uncleaned.csv")
df.info()
```

```python
missing_codes = [-9, -8, -7]
df = df.replace(missing_codes, np.nan)

print("\nAfter replacing special codes with NaN:")
df.info()
```

```python
missing_counts = df.isna().sum()
print("\nMissing values per column:")
print(missing_counts[missing_counts > 0].sort_values(ascending=False))
```

```python
# Remove any nulls from target, otherwise let XGBoost handle
df = df.dropna(subset=["RiskPerformance"])
print(df["RiskPerformance"].value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["RiskPerformance"] = le.fit_transform(df["RiskPerformance"])
print("\nEncoded RiskPerformance value counts:")
print(df["RiskPerformance"].value_counts())
print(dict(zip(le.classes_, le.transform(le.classes_))))
```

```python
df.to_csv("../data/processed/FICO_cleaned2.csv", index=False)
print("\nSaved cleaned data to ../data/processed/FICO_cleaned2.csv")
```
